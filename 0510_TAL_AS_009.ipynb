```python
from sklearn.datasets import load_breast_cancer
from sklearn.feature_selection import SelectKBest, f_classif # K es el parametro principal (kmeans / cantidad de , caracteristicas,cluster, clases)
```

```python
data= load_breast_cancer()
x= data.data
y= data.target
```

```python
selector= SelectKBest(score_func= f_classif, k=2)
```

```python
selector.fit_transform(x,y)
```

```python
indices= selector.get_support(indices=True)
print(indices)
print(data.feature_names[22])
print(data.feature_names[27])
```
