```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline #TUBERÍA PARA LOS DATOS DE ENTRAMIENTO
from sklearn.inspection import DecisionBoundaryDisplay #FRONTERA DE DECISIÓN
#otros modelos
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC

x,y= load_breast_cancer(return_X_y= True)
x=x[:, [22,27]]
print(load_breast_cancer().feature_names[22])
print(load_breast_cancer().feature_names[27])

x_train, x_test, y_train, y_test= train_test_split(x,y,test_size=0.2, random_state=42, stratify=y)
model= make_pipeline(StandardScaler(), SVC(kernel='linear', class_weight='balanced')) #tubería para escalar datos y definir modelo
model.fit(x_train, y_train)
print('Probando nuestras predicciones con la variables seleccionadas')
print(model.predict(x_test))
print(y_test)
DecisionBoundaryDisplay.from_estimator(model, x_train, response_method='predict', cmap='coolwarm', alpha=0.3)
plt.scatter(x_test[:,0], x_test[:,1], c= y_test, cmap='coolwarm', edgecolor= 'k', alpha=0.5)
plt.show()
```
