```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline #TUBERÍA PARA LOS DATOS DE ENTRAMIENTO
from sklearn.inspection import DecisionBoundaryDisplay #FRONTERA DE DECISIÓN
#otros modelos
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
```

```python
iris = load_iris()
x, y = iris.data, iris.target
```

```python
x,y= load_iris(return_X_y= True)
x = x[:, [2, 3]]
print("Característica X:", iris.feature_names[2])
print("Característica Y:", iris.feature_names[3])
```

```python
x_train, x_test, y_train, y_test= train_test_split(x,y,test_size=0.2, random_state=42, stratify=y)
model= make_pipeline(StandardScaler(), SVC(kernel='linear', class_weight='balanced')) #tubería para escalar datos y definir modelo
model.fit(x_train, y_train)
```

```python
print('Probando nuestras predicciones con la variables seleccionadas')
print(model.predict(x_test))
print("Reales:")
print(y_test)
DecisionBoundaryDisplay.from_estimator(model, x_train, response_method='predict', cmap="viridis", alpha=0.4)
plt.scatter(x_test[:,0], x_test[:,1], c= y_test, cmap='viridis', edgecolor= 'k', alpha=0.5)

plt.show()
```

```python
from sklearn.datasets import load_iris
from sklearn.feature_selection import SelectKBest, f_classif

iris = load_iris()
x, y = iris.data, iris.target

selector = SelectKBest(score_func=f_classif, k=2)
x_nuevo = selector.fit_transform(x, y)

indices = selector.get_support(indices=True)

print("Índices seleccionados:", indices)
for idx in indices:
    print(f"Índice {idx}: {iris.feature_names[idx]}")
```
