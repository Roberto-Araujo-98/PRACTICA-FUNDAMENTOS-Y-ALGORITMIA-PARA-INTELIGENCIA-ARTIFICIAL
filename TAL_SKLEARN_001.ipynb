```python
from sklearn.linear_model import LogisticRegression #clasificación binaria
from sklearn.datasets import load_breast_cancer #dataset (tumor(m/b))
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report #métricas clasificación
from sklearn.preprocessing import StandardScaler #escalar datos
```

```python
data = load_breast_cancer() #cargamos el dataset
x = data.data #solo datos
y = data.target #solo etiquetas
```

```python
print(" --- NOMBRES DE ETIQUETA --- \n",data.target_names)
print("--- CANTIDAD DE ETIQUETAS ---\n", data.target_names.shape)
print(" --- NOMBRES DE CARACTERÌSTICAS ---\n",data.feature_names)
print(" --- CANTIDAD DE CARECTERÍSTICAS ---\n",data.feature_names.shape)
```

```python
x.shape
```

```python
y.shape
```

```python
scaler = StandardScaler()
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
model =LogisticRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
print("Predicciones del modelo \n", y_pred)
print("Etiquetas reales", y_test)
```

```python
print(accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```
