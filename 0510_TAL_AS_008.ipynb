```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline #TUBERIA PARA LOS DATOS DE ENTRENAMIENTO
from sklearn.inspection import DecisionBoundaryDisplay #Frontera de Decision / genera un contenedor en 2d
#otros modelos
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
```

```python
x,y= load_breast_cancer(return_X_y=True) #  obtenemos x y y de un solo paso
x=x[:, [0,1]] # escojo 2 columnas q son 0 y 1 o primera y segunda columna
print(load_breast_cancer().feature_names[22])
print(load_breast_cancer().feature_names[27])
```

```python
x_train, x_test, y_train, y_test= train_test_split(x,y,test_size=0.2, random_state=42, stratify=y)
```

```python
model= make_pipeline(StandardScaler(),SVC(kernel='linear', class_weight='balanced'))
model.fit(x_train,y_train)
```

```python
print('Probando nuestras predicciones con la variable seleccionada')
print(model.predict(x_test))
print(y_test)
```

```python
DecisionBoundaryDisplay.from_estimator(model, x_train, response_method='predict', cmap='coolwarm', alpha=0.3)
plt.scatter(x_test[:,0], x_test[:,1], c= y_test, cmap='coolwarm', edgecolors='k')
```
