```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
ventas = pd.DataFrame({
    "vendedor": ["Pepe", "Sussie", "Pepe","Pepe", "Sussie"],
    "ventas": [1000, 500,600,2000,1500]
})
ventas
```

```python
ventas.groupby("vendedor")["ventas"].sum()
```

```python
#Venta màxima

venta_maxima = ventas["ventas"].max()
venta_maxima
```

```python
#Listas las venta de pepe

ventas[ventas["vendedor"] == "Pepe"]
```

```python
# Ventas màximas de Pepe y Sussie

ventas[ventas["vendedor"] == "Pepe"].max()

```

```python
ventas[ventas["vendedor"] == "Sussie"].max()
```

```python
#Ventas máximas de todos los vendedores
venta_max_vendedores = ventas.groupby("vendedor")["ventas"].max()
print(venta_max_vendedores)
print(venta_max_vendedores.index)
print(venta_max_vendedores.values)
plt.bar(venta_max_vendedores.index, venta_max_vendedores.values)
plt.title("Comparaciòn de ventas máximas", fontsize = 20, fontweight= "bold")
```

```python
#Promedio de venta de los los vendedores

promedio_ventas = ventas.groupby("vendedor")['ventas'].mean()
print(promedio_ventas)
print(promedio_ventas.index)
print(promedio_ventas.values)
plt.bar(promedio_ventas.index, promedio_ventas.values)
plt.title("Comparación de ventas promedio", fontsize = 20, fontweight= "bold")
```

```python
plt.bar(venta_max_vendedores.index, venta_max_vendedores.values, label ="Maximus")
plt.bar(promedio_ventas.index, promedio_ventas.values, label = "Promedius")
plt.legend()
```
