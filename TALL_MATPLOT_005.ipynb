```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips = sns.load_dataset("tips") #Cargamos el dataset de tips (propinas)
tips.head(10)
```

```python
data = tips[["tip", "sex"]]
```

```python
#Promedio de venta de los los vendedores
promedio_tips = data.groupby("sex")['tip'].mean()
print(promedio_tips)
```

```python
#Ventas máximas de todos los vendedores
max_tips = data.groupby("sex")["tip"].max()
print(max_tips)
```

```python
plt.bar(max_tips.index, max_tips.values, label ="Maximus", alpha=0.5)
plt.bar(promedio_tips.index, promedio_tips.values, label = "Promedius", alpha=0.5)
plt.title("Comparaciòn de propinas según el sexo", fontsize = 20, fontweight= "bold")
plt.legend()
```
