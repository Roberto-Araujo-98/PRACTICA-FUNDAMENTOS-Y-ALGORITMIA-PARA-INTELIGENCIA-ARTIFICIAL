```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
escuela= pd.DataFrame({
    'idescuela': [1,2,3],
    'nombre':['TI','NI','CS']
})
escuela
```

```python
carrera= pd.DataFrame({
    'idcarrera':[1,2,3,4,5,6],
    'nombre':['Redes','Ing. Software', 'Programación','Administración','Contabilidad','Farmacia'],
    'idescuela':[1, 1, 1, 2, 2, 3]
})
carrera
```

```python
vista_rok = carrera.merge(escuela, how="inner", on="idescuela")
vista_rok = vista_rok.rename(columns={"nombre_x":"carrera", "nombre_y": "escuela"})
vista_rok
```

```python
conteo = vista_rok.groupby("escuela")["carrera"].count()
print("---REPORTE ESTADÌSTICO 001...\n", conteo)
print("carrera", conteo.index)
print("conteo", conteo.values)
```

```python
plt.pie(conteo.values, labels=conteo.index, autopct="%1.1f%%")
plt.title("Reporte estadístico 001", fontsize=20, fontweight="bold", color="blue")
plt.legend()
plt.show()
plt.savefig("reporte001.png")
```
