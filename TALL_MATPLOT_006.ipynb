```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_iris
```

```python
iris= load_iris()
```

```python
df= pd.DataFrame(iris.data, columns= iris.feature_names)
df['species']= iris.target_names[iris.target]
df.head()
```

```python
promedio= df.groupby('species').mean()
promedio
```
