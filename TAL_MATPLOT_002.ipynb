```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
familia= pd.DataFrame({
    "idfamilia": [1,2,3],
    "nombre" : ["Bebidas","Lácteos","Escolar"]
    })
familia
```

```python
categoria = pd.DataFrame({ "idcategoria": [1,2,3,4],
                          "nombre":["Gaseosa", "Agua Mineral", "Leche", "Queso"],
                           "idfamilia": [1,1,2,2]
                           })
categoria
```

```python
vista_cat = categoria.merge(familia, how="inner", on="idfamilia")
vista_cat = vista_cat.rename(columns={"nombre_x":"categoria", "nombre_y": "familia"})
vista_cat
```

```python
vista_cat[["categoria", "familia"]]
```

```python
conteo = vista_cat.groupby("familia")["categoria"].count()
print("---REPORTE ESTADÌSTICO 001...\n", conteo)
print("categoria", conteo.index)
print("conteo", conteo.values)
```

```python
plt.pie(conteo.values, labels=conteo.index, autopct="%1.1f%%")
plt.legend()
plt.show()
```

```python
plt.bar(conteo.index, conteo.values, label=conteo.index, color = ["r","g"])
plt.show()
```

```python
colores =["purple", "yellow"]
for i, (nombre, valor) in enumerate(zip(conteo.index, conteo.values)):
    plt.bar(nombre,valor, label=nombre, color= colores[i])
plt.legend()
plt.show()
```

```python
fig, axes=plt.subplots(1,2)
axes[0].bar(conteo.index, conteo.values, color =colores)
axes[0].set_title("Gráfico 1")
axes[1].pie(conteo.values, labels=conteo.index, autopct="%1.1f%%")
axes[1].set_title("Gráfico 2")
plt.show()
```
